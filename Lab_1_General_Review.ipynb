```python

# First of all we define the function representing f(x)
def f(x):
    return x**3-2*x+2
# Now we calculate 1000 points of the function from -2 to 2. We choose n = 1000 arbitrarily.
x= np.linspace(-2,2,1000)
plt.plot(x, f(x))
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
```

**2. Compute analytically the points $x^∗$ that satisfy $f'(x) = 0$. Observe if the obtained result is congruent with the plot of the previous point.**

We have that $f'(x)=3x^2 -2$. Now we look for $f'(x)=0$, i.e., we have that the points are $x_1=-\sqrt\frac{2}{3}$ and $x_2=+\sqrt\frac{2}{3}$.
The obtained result is clearly congruent with the graph, and the points are marked in the graph. 


```python
#Now we plot the critical points on the plot.
def plot_f():
    plt.plot(x, f(x), label='f(x)')
    plt.xlabel('x')
    plt.grid(True)
    
x0= -np.sqrt(2/3)
x1=  np.sqrt(2/3)

plot_f()
plt.scatter(x0, f(x0), color='red', label="x* (critical points)")
plt.scatter(x1, f(x1), color='red')
plt.legend()
plt.show()
```

The points that satisfies $f'(x)=0$ are marked in red in the plot. We can observe that the analytical results are congruent with the plot. 

**3. We are now going to check which of the latter points $x^*$ are minima (or maxima). For that purpose, let us perform a 2nd order Taylor expansion around point $x^*$:**

$$ f(x^* + d) \approx f(x^*) + d f'(x^*) + \frac{1}{2} d^2 f''(x^*) \tag{1} $$

**where $d \in \mathbb{R}$ is the perturbation around $x^*$. Since we are dealing with a one-dimensional function, $f''(x^*)$ is a real number which may be positive or negative.**


The second derivative is $f''(x)=6x$. We evaluate the critical points to classify them:

* Case $x_0 = -\sqrt{\frac{2}{3}}$

* Case $x_1 = \sqrt{\frac{2}{3}}$

```python
def d2f(x):
    return 6*x
from IPython.display import display, Math

display(Math(r"f''(x_0)={:2.4f}".format(d2f(x0)))) #We use display so 
display(Math(r"f''(x_1)={:2.4f}".format(d2f(x1)))) #the output is in Latex
display(Math(
    r"\text{{We need to compute: }} "
    r"f(x_0)={:2.4f},\quad "
    r"f(x_1)={:2.4f},\quad "
    r"\frac{{1}}{{2}}f''(x_0)={:2.4f},\quad "
    r"\frac{{1}}{{2}}f''(x_1)={:2.4f}"
    .format(f(x0), f(x1), 0.5*d2f(x0), 0.5*d2f(x1))
))
```

**We are now going to focus on simple two-dimensional functions, $\mathbf{x} \in \mathbb{R}^2$, $\mathbf{x} = (x_1, x_2)^T$ (vectors are expressed column-wise). Let us begin with the next quadratic expression**

$$
f(\mathbf{x}) = x_1^2 + x_2^2

**Follow the next steps:**

**1. Plot this function. Please, note that this function has a minimum at $(x_1, x_2) = (0, 0)$.**

```python
def g(x_1, x_2):
  return x_1**2 + x_2**2

x_1 = np.linspace(-10,10,1000)
x_2 = np.linspace(-10,10,1000)
X_1, X_2 = np.meshgrid(x_1, x_2)
Z = g(X_1, X_2)

fig = plt.figure(figsize=(10,10))
ax = fig.add_subplot(projection='3d')
ax.plot_surface(X_1, X_2, Z, cmap='viridis')

ax.set_xlabel(r"$x_1$")
ax.set_ylabel(r"$x_2$")
ax.set_zlabel(r"$g(x_1, x_2)$")
plt.title(r"$g(x_1, x_2) = x_1^2 + x_2^2$")

plt.show()
```

**2. Analytically compute the gradient of the function, $\nabla f(\mathbf{x})$, and compute the point $\mathbf{x}^*$ at which $\nabla f(\mathbf{x}^*) = 0$.**



Given $f(x) = x_1^2 + x_2^2$, the gradient is defined as

$$
\nabla f(x) = \begin{bmatrix} \dfrac{\partial f}{\partial x_1} \\[6pt] \dfrac{\partial f}{\partial x_2} \end{bmatrix} = \begin{bmatrix} 2x_1 \\ 2x_2 \end{bmatrix}
$$


We compute the critical points by solving $\nabla f(x) = 0$,

$$
2x_1 = 0 \quad \Rightarrow \quad x_1 = 0
$$

$$
2x_2 = 0 \quad \Rightarrow \quad x_2 = 0
$$

Therefore, the critical point is

$$
x^* = (0, 0)
$$

**3. Let $\mathbf{d} \in \mathbb{R}^2$ be the perturbation around $\mathbf{x}^*$. The Taylor expansion, up to second order, of a function of several variables can be compactly expressed as**

$$
f(\mathbf{x}^* + \mathbf{d}) \approx f(\mathbf{x}^*) + \mathbf{d}^T \nabla f(\mathbf{x}^*) + \frac{1}{2} \mathbf{d}^T \nabla^2 f(\mathbf{x}^*)\, \mathbf{d} \tag{2}
$$

**Analyze the previous expression and be sure to understand the operations that are done at each of the terms.**

**Compute the Hessian matrix, $\nabla^2 f(\mathbf{x})$, at the point $\mathbf{x} = \mathbf{x}^*$. You should obtain**

$$
\nabla^2 f(\mathbf{x}^*) = \begin{pmatrix} 2 & 0 \\ 0 & 2 \end{pmatrix}
$$

**The latter matrix is giving us information about the shape of the quadratic approximation at $\mathbf{x} = \mathbf{x}^*$ in a similar way as has been done for the one dimensional case.**

**For the one-dimensional case it is easy to check if we have a minimum, $f''(x^*) > 0$, or a maximum, $f''(x^*) < 0$. For a higher dimensional problem we are sure that the quadratic approximation is convex and that we have a minimum if**

$$
\mathbf{d}^T \nabla^2 f(\mathbf{x}^*)\, \mathbf{d} > 0 \quad \mathbf{d} \neq \mathbf{0} \tag{3}
$$

**We have a maximum if**

$$
\mathbf{d}^T \nabla^2 f(\mathbf{x}^*)\, \mathbf{d} < 0 \quad \mathbf{d} \neq \mathbf{0} \tag{4}
$$

**The previous conditions can be verified by computing the eigenvalues of $\nabla^2 f(\mathbf{x}^*)$. If all eigenvalues are strictly positive, equation (3) is satisfied. If all eigenvalues are strictly negative, equation (4) is satisfied. For this example, which are the eigenvalues of the Hessian matrix? Do we have a minimum or a maximum at $\mathbf{x}^*$?**


The Hessian is the matrix of second partial derivatives, i.e,

$$
H(x) = \begin{bmatrix} \dfrac{\partial^2 f}{\partial x_1^2} & \dfrac{\partial^2 f}{\partial x_1 \partial x_2} \\[6pt] \dfrac{\partial^2 f}{\partial x_2 \partial x_1} & \dfrac{\partial^2 f}{\partial x_2^2} \end{bmatrix}.
$$

Calculating each term for $f(x) = x_1^2 + x_2^2$, we have that

$$
\frac{\partial^2 f}{\partial x_1^2} = 2, \quad \frac{\partial^2 f}{\partial x_2^2} = 2, \quad \frac{\partial^2 f}{\partial x_1 \partial x_2} = 0.
$$

As the function $f$ has continous second partial derivatives, we have by Schwarz that $\frac{\partial^2 f}{\partial x_1 \partial x_2} = \frac{\partial^2 f}{\partial x_2 \partial x_1}$.

Therefore, the Hessian is

$$
H(x^*) = \begin{bmatrix} 2 & 0 \\ 0 & 2 \end{bmatrix}.
$$


We find the eigenvalues by solving $\det(H - \lambda I) = 0$, i.e.,

$$(2-\lambda)^2 = 0 \quad \Rightarrow \quad \lambda_1 = \lambda_2 = 2$$

Both eigenvalues are $\lambda_1 = \lambda_2 = 2 > 0$.

Since both eigenvalues are strictly positive, $H(x^*)$ is positive definite, so $x^*=(0,0)$ is a minimum (in fact, the global minimum, since $f$ is convex everywhere).

**3. Numerically compute an approximation of the points $\mathbf{x}^*$ at which $\nabla f(\mathbf{x}^*)=0$**

**a) Evaluate $\|\nabla f(x)\|^2$ over the previous range using a step of e.g. 0.005 or smaller if you prefer (but not too small!).**

**You may create a matrix that stores all these values to be able to analyze them in the next steps.**

```python
def gradf_norm(x_1, x_2):
    gradf_1=8*x_1-8.4*x_1**3+2*x_1**5+x_2
    gradf_2=x_1-8*x_2+16*x_2**3
    return gradf_1**2+gradf_2**2
#We note that return np.sqrt(gradf_1**2+gradf_2**2)**2 is equivalent to gradf_1**2+gradf_2**2. So we use the simpler version.

x_1 = np.arange(-2, 2+0.005, 0.005)
x_2 = np.arange(-1, 1+0.005, 0.005)
X_1, X_2 = np.meshgrid(x_1, x_2, indexing="ij")
#We add indexing "ij" because it will make more natural how we explore the matrix on the next cell
#That's because in Z the first component fixes x_2 and the second one fixes x_1 if we don't add indexing="ij"
#As we added it, the first component fixes x_1 and the second one fixes x_2, that's a more natural way to understand it.

Z = gradf_norm(X_1, X_2)
Z.shape
```

```python
aux_x_barret=[]
aux_value=[]
for i in range(1, Z.shape[0]-1):
    for j in range (1, Z.shape[1]-1):
        if((Z[i,j]<Z[i-1,j]) and (Z[i,j]<Z[i-1, j-1]) and (Z[i,j]<Z[i, j-1]) and (Z[i,j]<Z[i-1, j+1]) and (Z[i,j]<Z[i+1, j-1]) and (Z[i,j]<Z[i+1, j+1]) and (Z[i,j]<Z[i, j+1]) and (Z[i,j]<Z[i+1, j])):
            aux_x_barret.append([-2+0.005*i, -1+0.005*j])  
            aux_value.append([float(Z[i, j])])                 
```

4. Compute the analytical expression of the Hessian of $f(x_1, x_2)$ and evaluate it at the values $\tilde{\mathbf{x}}$ you have found. What kind of information is giving you the Hessian? Does it correspond to a minimum (the Hessian is positive definite)? To a maximum (the Hessian is negative definite)? Or may be a saddle point? You may use the `eigvals` function of Python to compute the eigenvalues of the Hessian matrix (no need to compute them analytically). Take into account that there may be several minima, maxima and saddle points for the function you are analyzing.

We have that the Hessian of $f(x_1, x_2)$ is
$$
H_f(x) = \nabla^2 f(x) =
\begin{bmatrix}
\displaystyle \frac{\partial^2 f}{\partial x_1^2}
&
\displaystyle \frac{\partial^2 f}{\partial x_1 \partial x_2}
\\[8pt]
\displaystyle \frac{\partial^2 f}{\partial x_2 \partial x_1}
&
\displaystyle \frac{\partial^2 f}{\partial x_2^2}
\end{bmatrix}
$$
So, 
$$
H_f(x) =
\begin{bmatrix}
8 - 25.2x_1^2 + 10x_1^4 & 1 \\[6pt]
1 & -8 + 48x_2^2
\end{bmatrix}
$$

```python
def Hessian_f(x_1,x_2):
    return np.array([
        [8 - 25.2 * x_1**2 + 10 * x_1**4, 1],
        [1, -8 + 48 * x_2**2]
    ])
#We add a tolerance tol instead of using 0, to avoid numerical errors. The tol we set is arbitrary.
tol = 1e-6
for x_1, x_2 in aux_x_barret:
    Hess=Hessian_f(x_1, x_2)
    eigenvalues = np.linalg.eigvals(Hess)
    if (eigenvalues > tol).all():
        print("The point ({}, {}) is a local minimum and its Hessian is \n {}.".format(x_1, x_2, Hess))

    elif (eigenvalues < -tol).all():
        print("The point ({}, {}) is a local maximum and its Hessian is \n {}.".format(x_1, x_2, Hess))
    elif (eigenvalues > tol).any() and (eigenvalues < -tol).any():
        print("The point ({}, {}) is a saddle point and its Hessian is \n {}.".format(x_1, x_2, Hess))
    else:
        print("We need more information to determine the nature of the point ({}, {}).".format(x_1, x_2))
    print("The eigenvalues of the Hessian are: {}".format(eigenvalues))
```
